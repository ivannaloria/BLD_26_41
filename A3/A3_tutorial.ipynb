## Exercise 1: Open an IFC model
`ifcopenshell` is the Python library that reads IFC files. `ifcopenshell.open()` loads the whole building into a variable we call `model`.

```python
import ifcopenshell

model = ifcopenshell.open(IFC_FILE)
print("Model opened. IFC schema:", model.schema)
```

**Expected output:** `Model opened. IFC schema: IFC4X3` for the supplied file. The schema names the IFC data structure used by this model. `import` makes the library available; `model` holds the opened IFC data.

If you see `FileNotFoundError` or a path-related error, check the filename and working folder in Step 0. If Python says `model` is not defined later, run this cell first. Opening and querying the file does not save changes to it.

## Exercise 2: Identify building elements

**Purpose:** discover which walls are recorded in the building data.

Every element in an IFC model has a class, such as `IfcWall`, `IfcSlab`, or `IfcDoor`. `model.by_type("IfcWall")` returns the walls (including subclasses); `len()` counts the results. `walls` is the name we give this collection.

```python
walls = model.by_type("IfcWall")
print("Number of walls:", len(walls))
```

**Expected output:** `Number of walls: 149` for the supplied model. This counts IFC elements classified as walls, rather than shapes selected in the 3D view. A different model can have different counts or no walls at all. A zero count is a finding, not automatically a Python error.

## Exercise 3: Retrieve element information

**Purpose:** identify individual walls, rather than only count them.

- **Name:** a human-readable label from the modelling software. Names can repeat or be missing (`None`).
- **GlobalId:** an identifier intended to uniquely identify an IFC element. Use it to compare elements between tools opening this same model.

`walls[:5]` means “up to the first five walls”. A `for` loop repeats the indented `print()` line for each wall; `wall.Name` and `wall.GlobalId` read its attributes. Keep the indentation below `for`.

```python
for wall in walls[:5]:
    print(wall.Name, wall.GlobalId)
```

**Expected output:** five rows, each containing a name and GlobalId. For the supplied file, the first wall is named `Basic Wall:BA_Heavy Weight_Concrete_200mm:1794121` and has GlobalId `2xflkD7IDEMgp51ZbDNMUT`. A printed `None` means no name was supplied. These are identities, not dimensions or coordinates.

## Exercise 5: Connecting Python and Blender/Bonsai

**Purpose:** show that the visible building and the IFC data describe the same elements.

The cell below uses the model already opened in Exercise 1. `walls[0]` selects the first wall (Python starts counting at zero); `element.is_a()` reports its IFC class. The `if walls` check avoids selecting from an empty collection when you try another model.

```python
if walls:
    element = walls[0]
    print("Name:", element.Name)
    print("IFC class:", element.is_a())
    print("GlobalId:", element.GlobalId)
else:
    print("No walls found. Try a class present in your model and rerun Exercises 2–5.")
```
