## Exercise 1: Open an IFC model
`ifcopenshell` is the Python library that reads IFC files. `ifcopenshell.open()` loads the whole building into a variable we call `model`.

```python
import ifcopenshell

model = ifcopenshell.open(IFC_FILE)
print("Model opened. IFC schema:", model.schema)
```

## Exercise 2: Count building elements
Every element in an IFC model has a class, like `IfcWall`, `IfcSlab` or `IfcDoor`. `model.by_type()` returns all elements of one class as a list, and `len()` counts them.

```python
walls = model.by_type("IfcWall")
print("Number of walls:", len(walls))
```

## Exercise 3: Retrieve element information
Each element has attributes. Two useful ones are:
- **Name**: the name given in the modelling software
- **GlobalId**: a unique ID that never repeats in the model

`walls[:5]` means "the first 5 walls".

```python
for wall in walls[:5]:
    print(wall.Name, wall.GlobalId)
```

## Exercise 5: Extract quantities
For a cost estimation we need more than names: we need **quantities** like areas and volumes. IFC stores them in quantity sets. For walls, the set is called `Qto_WallBaseQuantities`.

`get_psets(..., qtos_only=True)` returns only the quantity sets of an element. If a value prints as `None`, that quantity was not exported in the model.

```python
import ifcopenshell.util.element as util

for wall in walls[:5]:
    qto = util.get_psets(wall, qtos_only=True).get("Qto_WallBaseQuantities", {})
    print(wall.Name, "| Net side area (m2):", qto.get("NetSideArea"))
```
